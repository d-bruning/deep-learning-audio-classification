```python
from pathlib import Path

import numpy as np
import pandas as pd


MODEL_NAME = "vit"

MODEL_DIR = (
    Path(MODEL_NAME)
)

SUMMARY_FILE = (
    MODEL_DIR /
    "duration_results" /
    f"{MODEL_NAME}_experiment_summary_all.csv"
)

HISTORY_DIR = (
    MODEL_DIR /
    "histories"
)


summary_df = pd.read_csv(
    SUMMARY_FILE
)


for idx, row in summary_df.iterrows():

    duration = row["duration"]
    seed = row["seed"]

    history_file = (
        HISTORY_DIR /
        f"training_history_{duration}_seed_{seed}.csv"
    )

    if not history_file.exists():

        print(
            f"Missing history: "
            f"{history_file}"
        )

        continue

    history_df = pd.read_csv(
        history_file
    )

    final_train_accuracy = (
        history_df["accuracy"]
        .iloc[-1]
    )

    final_train_loss = (
        history_df["loss"]
        .iloc[-1]
    )

    final_val_accuracy = (
        history_df["val_accuracy"]
        .iloc[-1]
    )

    final_val_loss = (
        history_df["val_loss"]
        .iloc[-1]
    )

    initial_val_accuracy = (
        history_df["val_accuracy"]
        .iloc[0]
    )

    epochs_completed = len(
        history_df
    )

    best_epoch = row[
        "best_epoch"
    ]

    accuracy_gain = (
        row["best_val_accuracy"]
        - initial_val_accuracy
    )

    generalization_gap = (
        final_train_accuracy
        - final_val_accuracy
    )

    best_epoch_fraction = (
        best_epoch
        / epochs_completed
    )

    overfitting_flag = (
        generalization_gap > 0.20
    )

    summary_df.loc[
        idx,
        "final_train_accuracy"
    ] = final_train_accuracy

    summary_df.loc[
        idx,
        "final_train_loss"
    ] = final_train_loss

    summary_df.loc[
        idx,
        "final_val_accuracy"
    ] = final_val_accuracy

    summary_df.loc[
        idx,
        "final_val_loss"
    ] = final_val_loss

    summary_df.loc[
        idx,
        "accuracy_gain"
    ] = accuracy_gain

    summary_df.loc[
        idx,
        "generalization_gap"
    ] = generalization_gap

    summary_df.loc[
        idx,
        "best_epoch_fraction"
    ] = best_epoch_fraction

    summary_df.loc[
        idx,
        "overfitting_flag"
    ] = overfitting_flag


summary_df.to_csv(
    SUMMARY_FILE,
    index=False
)

print(
    "experiment_summary_all.csv updated"
)
```
